## 3. Modulo `%`

### Expectation
You should know that `%` gives the **remainder** after division and can be used to check divisibility.

For example, a number is divisible by 5 when `number % 5 == 0`.

**Resource:** [Python arithmetic operators](https://docs.python.org/3/reference/expressions.html#binary-arithmetic-operations)


```python
number = 28
print(number % 5)

if number % 2 == 0:
    print("even")
else:
    print("odd")
```

## 9. Booleans and logical operators

### Expectation
You should be able to evaluate conditions using:
- `and` - both conditions must be True
- `or` - at least one condition must be True
- `not` - reverses True/False

**Resource:** [Python Boolean operations](https://docs.python.org/3/reference/expressions.html#boolean-operations)


```python
age = 20
has_id = False

print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
```

## 10. `if`, `elif`, and `else`

### Expectation
You should be able to trace a conditional and determine which branch runs.

Python checks conditions from top to bottom and runs the first matching branch.

**Resource:** [Python `if` statements](https://docs.python.org/3/tutorial/controlflow.html#if-statements)


```python
temperature = 55
raining = True

if temperature > 70 and not raining:
    print("Picnic")
elif temperature > 50:
    print("Bring a jacket")
else:
    print("Stay warm")
```

## 11. Translating math into Python

### Expectation
You should be able to translate standard math notation into valid Python syntax.

Common translations:
- x² -> `x**2`
- 3x -> `3*x`
- sin(x) -> `math.sin(x)`
- cos(x) -> `math.cos(x)`
- √x -> `math.sqrt(x)`

Remember to use `*` for multiplication.

**Resource:** [Python `math` module](https://docs.python.org/3/library/math.html)


```python
import math

x = 2
h = x**2 - 4*x + math.sin(x)
print(h)
```
