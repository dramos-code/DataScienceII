## 8. Why capitalization matters

Python treats these as different strings:

```python
"Employed"
"employed"
"EMPLOYED"
```

One simple cleaning tool is `.capitalize()`.

`.capitalize()` makes:
- the **first letter uppercase**
- the rest of the word lowercase

Example:

```python
"EMPLOYED".capitalize()
```

becomes:

```text
Employed
```

**Resource:** [Python string method documentation](https://docs.python.org/3/library/stdtypes.html#str.capitalize)


````python
employment = "EMPLOYED"

employment = employment.capitalize()

print(employment)
````

## 9. Clean and Check Employment Status
